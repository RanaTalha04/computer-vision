```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("../images/cat1.jpg")
print(type(image))
print(image.shape)
```

```python
plt.imshow(image)
plt.axis("off")
plt.show()
```

```python
crrt_color_imge = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(crrt_color_imge)
plt.axis("off")
plt.show()
```

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

plt.imshow(gray_image, cmap="gray")
plt.axis('off')
plt.show()
```
